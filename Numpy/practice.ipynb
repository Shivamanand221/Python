```python
X = np.array([
    [20, 50000],
    [25, 60000],
    [30, 75000],
    [35, 90000],
    [40, 120000]
])

print(X[X[:, 1]>70000])
```

```python
print(np.mean(X[0:,0:1]))
print(np.mean(X[0:,1:]))
print(np.mean(X, axis=0))
print(np.mean(X, axis=1))
```

```python
X = np.array([
    [10, 100],
    [20, 200],
    [30, 300]
])

feature_means= np.mean(X, axis=0)
centered_X= X-feature_means

print(feature_means)
print(centered_X)
```

```python
data = np.array([
    [10, 20, 30, 40],
    [15, 25, 35, 45],
    [20, 30, 40, 50],
    [25, 35, 45, 55]
])

print(
    np.shape(data),
    np.mean(data, axis=0),
    np.max(data, axis=0),
    data[data[:,3]>45],
    data-np.mean(data, axis=0),
    np.std(data, axis=0),
    sep='\n'
)
```

```python
X = np.array([
    [10, 100, 1],
    [20, 200, 2],
    [30, 300, 3],
    [40, 400, 4],
    [50, 500, 5]
])

m= np.mean(X, axis=0)
s= np.std(X, axis=0)
z= (X-m)/s
print(
    m,
    s,
    z,
    np.mean(z, axis=0),
    np.std(z, axis=0),
    sep='\n'
)
```
