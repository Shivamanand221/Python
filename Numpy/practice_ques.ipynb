```python
import numpy as np
```

```python
arr1= np.array([2,4,6,8])
print(arr1*3)
```

```python
arr2= np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
print(arr2[1:,0:2])
```

```python
arr3= np.arange(1,13)
arr8= arr3.reshape(3,4)
print(arr8)
print(arr8[:,2])
```

```python
arr4= np.arange(1,10,2)
print(arr4)
```

```python
arr5= np.array([10,20,30,40,50])
print(arr5[arr5>25])

print(
    np.mean(arr5),
    np.argmax(arr5),
    np.cumsum(arr5)
)

print(np.where(arr5>=30,1,0))
```

```python
arr6=np.array([
    [1,2],
    [3,4]
])
arr7=np.array([
    [5,6],
    [7,8]
])
print(np.concatenate((arr6,arr7)))
print(np.stack((arr6, arr7), axis=1))
```

```python
arr9= np.array([
    [10,20,30],
    [40,50,60]
])

print(
    np.sum(arr9, axis=0),
    np.sum(arr9, axis=1),
    sep='\n'
)
```
