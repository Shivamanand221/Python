```python
arr3= np.arange(1,13)
arr8= arr3.reshape(3,4)
print(arr8)
print(arr8[:,2])
```

```python
arr4= np.arange(1,10,2)
print(arr4)
```

```python
arr5= np.array([10,20,30,40,50])
print(arr5[arr5>25])

print(
    np.mean(arr5),
    np.argmax(arr5),
    np.cumsum(arr5)
)

print(np.where(arr5>=30,1,0))
```

```python
arr6=np.array([
    [1,2],
    [3,4]
])
arr7=np.array([
    [5,6],
    [7,8]
])
print(np.concatenate((arr6,arr7)))
print(np.stack((arr6, arr7), axis=1))
```

```python
arr9= np.array([
    [10,20,30],
    [40,50,60]
])

print(
    np.sum(arr9, axis=0),
    np.sum(arr9, axis=1),
    sep='\n'
)
```

```python
marks = np.array([35, 72, 48, 91, 63, 27, 84])
print(marks[(marks>50) & (marks<90)])
```

```python
print(np.where(marks>=40,1,0))
```

```python
scores = np.array([45, 67, 89, 32, 76, 91, 54])
print(
    np.mean(scores),
    np.median(scores),
    np.min(scores),
    np.max(scores),
    np.argmax(scores),
    np.std(scores)
)
```
