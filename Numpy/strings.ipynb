```python
import numpy as np
```

```python
names= np.array(["ayush", "rahul", "aman"])
print(names)

s1= np.char.upper(names)
s2= np.char.lower(names)

print(s1,s2, sep='\n')

s3= np.char.capitalize(names)
print(s3)

s4= np.char.str_len(names)
print(s4)
```

```python
first= np.array(["rahul", "aman"])
last= np.array(["sharma", "kumar"])

add1= np.char.add(first, last)
print(add1)

add2= np.char.add(
    np.char.add(first, " "),
    last
    )

print(add2)

c1= np.char.capitalize(add2)
print(c1)
```

```python
# Datetime

# Y  → year
# M  → month
# D  → day
# h  → hour
# m  → minute
# s  → second
# ms → millisecond
```

```python
date= np.datetime64("2026-09-06")
print(date)

date= date+ np.timedelta64(10,"D")
print(date)

start= np.datetime64("2026-09-01")
end= np.datetime64("2026-09-07")

print(end-start)
```

```python
names = np.array(["apple", "banana", "mango"])

print(np.char.upper(names))
print(np.char.str_len(names))

date = np.datetime64("2026-09-06")
print(date + np.timedelta64(7, "D"))
```

```python
dates= np.array([
    "2026-09-01",
    "2026-09-06",
    "2026-09-11"
], dtype="datetime64[D]")

print(dates)

print(dates + np.timedelta64(5, "D"))
print(dates + np.timedelta64(2, "D"))
print(dates + np.timedelta64(170, "D"))
```
