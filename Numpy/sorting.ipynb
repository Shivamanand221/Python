```python
import numpy as np
```

```python
a = np.array([40, 10, 30, 20, 50])
b= np.sort(a)
print(np.sort(a))
print(b)

print(np.sort(a)[::-1])
print(np.argsort(a))
```

```python
names= np.array(["A", "B", "C", "D"])
scores= np.array([80, 50, 90, 70])

idx= np.argsort(scores)
print(idx)

print(names[idx])
```

```python
a= np.array([
    [30, 10, 20],
    [60, 40, 50]
])

print(np.sort(a))
print(np.sort(a, axis=0))
print(np.sort(a, axis=1))
```

```python
a2 = np.array([
    [30, 50, 10],
    [20, 40, 60]
])
print(np.sort(a2))
print(np.sort(a2, axis=0))
print(np.sort(a2, axis=1))
```
