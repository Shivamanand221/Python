```python
import numpy as np
```

```python
print(np.random.rand())
print(np.random.rand(5))
print(np.random.rand(2, 3))
print(np.random.randint(1,10))
print(np.random.randint(1,10, 5))
print(np.random.randint(1,10, (2,3)))
print(np.random.random(5))
print(np.random.randn(5)) #mean=0, std=1
print(np.random.normal(100, 15, 5)) #mean=100, std=15, 5 values
print(np.random.seed(42))

a = np.array([10, 20, 30, 40, 50])
np.random.choice(a)
```

```python
a= np.array([10, 20, 30, 40, 50])

print(
    np.random.rand(),
    np.random.rand(7),
    np.random.rand(3,2),
    np.random.randint(1, 10, 5),
    np.random.randint(1, 20, (3,3)),
    np.random.randn(5),
    np.random.random(5),
    np.random.normal(100, 15, 5), #mean, std, no. of values
    np.random.choice(a, 7),
    np.random.choice(a, 3, replace= False),# cannot repeat values
    sep='\n'
)

np.random.seed(10)
print(np.random.randint(1, 100, 5))

np.random.shuffle(a)
print("a=", a)

c= np.random.permutation(a)
print("c=", c)

rng= np.random.default_rng(42)
print(rng.random(3))
```
