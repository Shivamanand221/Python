```python
import numpy as np
```

```python
A= np.array([
    [1,2],
    [3,4]
])

B= np.array([
    [5,6],
    [7,8]
])

print(A*B) #element wise multiplication
print(A@B) #matrix multiplication
print(np.matmul(A,B)) #matrix multiplication
print(np.dot(A,B)) 
```

```python
A1= np.array([
    [1,2,3],
    [4,5,6]
])

B1= np.array([
    [10,20],
    [30,40],
    [50,60]
])

print(A1@B1)
```

```python
A2= np.array([
    [1,2,3],
    [4,5,6]
])
B2=A2.T #transpose

print(A2, B2, sep='\n')
#print(A2*B2) not possible
print(A2@A2.T)
print(A2@B2)
```

```python
A3= np.array([
    [1,2],
    [3,4]
])

A3_inv=np.linalg.inv(A3) #inverse
print(A3_inv)

I= A3@A3_inv
print(I)

A3_det= np.linalg.det(A3) #determinant= det(A) = (a × d) - (b × c)
print(A3_det)

A3_eig= np.linalg.eig(A3)
print(A3_eig)

A3_values, A3_vectors= np.linalg.eig(A3)
print(A3_values, A3_vectors, sep='\n')
```

```python
A4 = np.array([
    [3, 0],
    [0, 2]
])

values, vectors = np.linalg.eig(A4)

print(values)
print(vectors)
```

```python
A5= np.array([
    [2,0],
    [0,5]
])

values, vectors= np.linalg.eig(A5)
print(values, vectors, sep='\n')
```

```python
#equation
#2x + y = 5
#x + 3y = 6

A6= np.array([
    [2,1],
    [1,3]
])

b= np.array([5,6])

x= np.linalg.solve(A6, b) #np.linalg.inv(A6) @ b
print(x)


A7= np.array([
    [1,1],
    [1,-1]
])

b1= np.array([10,2])

x1= np.linalg.solve(A7, b1)
print(x1)
```
