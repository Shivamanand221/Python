```python
import numpy as np
```

```python
A= np.array([
    [1,2],
    [3,4]
])

B= np.array([
    [5,6],
    [7,8]
])

print(A*B) #element wise multiplication
print(A@B) #matrix multiplication
print(np.matmul(A,B)) #matrix multiplication
print(np.dot(A,B)) 
```

```python
A1= np.array([
    [1,2,3],
    [4,5,6]
])

B1= np.array([
    [10,20],
    [30,40],
    [50,60]
])

print(A1@B1)
```

```python
A2= np.array([
    [1,2,3],
    [4,5,6]
])
B2=A2.T #transpose

print(A2, B2, sep='\n')
#print(A2*B2) not possible
print(A2@A2.T)
print(A2@B2)
```
