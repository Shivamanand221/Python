```python
import numpy as np
```

```python
a= np.array([10,20,30,40,50])
np.save("data.npy", a)
```

```python
b= np.load("data.npy")
print(b)
```

```python
a1= np.array([10,20,30,40,50])
np.savetxt("data.txt", a1)
```

```python
b1= np.loadtxt("data.txt")
print(b1)
```

```python
a = np.array([
    [1, 2],
    [3, 4]
])

np.save("matrix.npy", a)

b= np.load("matrix.npy")
print(b)
```

```python
a = np.array([
    [1, 2, 3],
    [4, 5, 6]
], dtype=np.int32)

np.save("data.npy", a)

b = np.load("data.npy")

print(b)
print(b.shape)
print(b.dtype)
```

```python
a= np.array([1,2,3])
b= np.array([10,20,30])

np.savez("data.npz", arr1=a, arr2=b)

data= np.load("data.npz")
print(data,
    data["arr1"],
    data["arr2"],
    sep='\n'
)
```

```python
a= np.array([1,2,3])
b= np.array([11,22,33])

np.savez("data.npz", a, b)

data1= np.load("data.npz")

print(data1)

a_loaded= data1["arr_0"]
b_loaded= data1["arr_1"]

print(
    a_loaded,
    b_loaded,
    sep='\n'
)
```

```python
np.savetxt("data.csv", a, delimiter=",")
a= np.loadtxt("data.csv", delimiter=",")

print(a)
```

```python
names = np.array(["Ayush", "Rahul", "Aman"])
ages = np.array([22, 24, 21])
marks = np.array([85, 90, 78])

data= np.column_stack((names, ages, marks))

np.savetxt("students.csv", data, delimiter=",", fmt="%s")

student= np.loadtxt("students.csv", delimiter=",", dtype=str)
print(student)

names_loaded= student[:,0]
ages_loaded= student[:,1].astype(int)
marks_loaded= student[:,2].astype(np.int32)

print(names_loaded)
print(ages_loaded.dtype)
print(marks_loaded.dtype)
```
