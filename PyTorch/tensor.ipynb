```python
x3 = torch.tensor(
    [1, 2, 3],
    dtype= torch.float32
)

print(x3)
print(x3.dtype, '\n')

y = torch.tensor([1, 2, 3])
print(y.shape)
print(y.dtype, '\n')

y1 = y.float()
print(y1)
print(y1.dtype, '\n')

y2 = y.to(torch.float32)
print(y2)
print(y2.dtype)
```

```python
a4 = torch.tensor([1, 2, 3])
b4 = torch.tensor([1.0, 2.0, 3.0])
c4 = torch.tensor([1, 2, 3], dtype=torch.float32)

print(a4)
print(a4.dtype, '\n')

print(b4)
print(b4.dtype, '\n')

print(c4)
print(c4.dtype)
```

```python
x4 = torch.tensor([1, 2, 3])

print(x4)
print(x4.dtype, '\n')

y3 = x4.float()
print(y3)
print(y3.dtype, '\n')

y4 = x4.to(torch.float32)
print(y4)
print(y4.dtype, '\n')

x4 = x4.float()
print(x4)
print(x4.dtype, '\n')

x5 = torch.tensor([1.5, 2.7, 3.9])
print(x5)
print(x5.type, '\n')

y5 = x5.int()
print(y5)
print(y5.dtype, '\n')

x6 = torch.tensor([1, 2, 3], dtype=float)
print(x6, '\n')

y6 = torch.tensor([1.5, 2.6, 3.6], dtype=int)
print(y6)
```
