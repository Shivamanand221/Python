```python
import numpy as np
import torch
```

```python
arr= np.array([1,2,3])
x= torch.from_numpy(arr)

print(
    arr,
    x,
    type(arr),
    type(x),
    sep='\n'
)
```

```python
arr= np.array([1,2,3])
x= torch.from_numpy(arr)

arr[0]=100
x[1]=200

print(
    arr,
    x,
    sep='\n'
)
```

```python
x= torch.tensor([10,20,30])

arr= x.numpy()

arr[0]=99
x[1]=100

print(
    x,
    arr,
    type(x),
    type(arr),
    sep='\n'
)
```

```python
arr= np.array([1,2,3])

x=torch.tensor(arr) #just a copy
x1= torch.from_numpy(arr) #shared memory

x[0]=99
x1[1]=100

print(
    arr,
    x,
    x1,
    sep='\n'
)
```
