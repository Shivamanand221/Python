```python
import torch
```

```python
x = torch.tensor([10, 20, 30])
y = torch.tensor([1, 2, 3])

print(x + y)
print(torch.add(x, y))

print(x-y)
print(torch.sub(x, y))

print(x * y)
print(torch.mul(x, y))

print(x/y)
torch.div(x, y)
```

```python
x = torch.tensor([2, 3, 4])

print(x ** 2)
print(torch.pow(x,2))
```

```python
x = torch.tensor([10, 20, 30])

print(x.sum())
print(torch.sum(x))

y = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print(y.sum(dim=0))
print(torch.sum(y,dim=0))
```

```python
x = torch.tensor([10, 20, 30])

print(x.float().mean())

y = torch.tensor([10, 20, 30, 40])

print(y.float().mean())
print(torch.mean(y.float()))

x = torch.tensor([10, 30, 5, 20, 15])

print(x.min())
print(x.max())

y = torch.tensor([
    [10, 20, 30],
    [40, 5, 60]
])

print(y.min(dim=0))
print(y.min(dim=1))

x = torch.tensor([15, 3, 27, 8, 20])

print(x.min())
print(x.max())
print(x.argmin())
print(x.argmax())
```

```python
x = torch.tensor([
    [10, 20, 30],
    [40, 5, 60]
])

print(x.min())
print(x.argmin())

print(x.max(dim=0))
print(x.argmax(dim=0))

result= x.min(dim=0)

print(result.values)
print(result.indices)
```

```python
x = torch.tensor([-10, -5, 0, 5, 10])

print(torch.abs(x)) #abs give absolute values

y = torch.tensor([-20, 7, -3, 0, 15])

print(y.abs())
print(torch.abs(y))
```

```python
x = torch.tensor([4, 9, 16])

print(torch.sqrt(x))

x = torch.tensor([1.0, 25.0, 36.0, 49.0])

print(torch.sqrt(x))
print(x.sqrt())
```

```python
x = torch.tensor([0.0, 1.0, 2.0])
print(torch.exp(x)) #exp gives e^x

y = torch.tensor([1.0, 2.7183, 7.3891])
print(torch.log(y)) #gives ln(y)

x = torch.tensor([1.0, 4.0, 9.0, 16.0])

print("sqrt:", torch.sqrt(x))
print("exp:", torch.exp(x))
print("log:", torch.log(x))
```

```python
x = torch.tensor([1.2, 2.5, 3.7, -1.2, -1.8])

print(torch.round(x)) #round values
print(torch.floor(x)) #gives nearest lesser integer
print(torch.ceil(x)) #gives nearest greater integer
```

```python
x = torch.tensor([10, 20, 30])

print(x > 15)
print(x < 25)
print(x == 20)
print(x != 20)

a = torch.tensor([1, 2, 3])
b = torch.tensor([1, 2, 3])
c = torch.tensor([1, 2, 4])

print(torch.equal(a, b))
print(torch.equal(a, c))
```

```python
x= torch.tensor([5,15,25,35,45])
print(torch.clamp(x, min=10, max=30)) #gives either below min. or above max

y= torch.tensor([2,3,4])
print(torch.prod(y))

z = torch.tensor([10.0, 20.0, 30.0])
print(z.var()) #var → variance
print(z.std()) #std → standard deviation
```

```python
x = torch.tensor([3.0, 4.0])
print(torch.norm(x)) #pythagorous theorum

y = torch.tensor([1.0, 2.0, 3.0])
z = torch.tensor([4.0, 5.0, 6.0])

print(torch.dot(y, z))

A = torch.tensor([
    [1, 2],
    [3, 4]
])

B = torch.tensor([
    [5, 6],
    [7, 8]
])

print(A @ B)
print(torch.matmul(A, B))
```

```python
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print(x.shape)

y = x.transpose(0, 1) #swap dimentions 0 and 1

print(y)
print(y.shape)

print(x.T)
```

```python
x = torch.tensor([True, False, False])

print(x.any())
print(x.all())

y= torch.tensor([10,20,30])

print((y>15).any())
print((y>15).all())
```

```python
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print(x.transpose(0, 1))
print(x.T)

condition = x > 3

print(condition)
print(condition.any())
print(condition.all())
```

```python
x = torch.tensor([1, 2, 3, 4])

print(x.cumsum(dim=0)) #cumulative sum
print(x.cumprod(dim=0))


z = torch.tensor([2, 3, 4, 5])

print("prod:", z.prod())
print("cumsum:", z.cumsum(dim=0))
print("cumprod:", z.cumprod(dim=0))
```

```python
x= torch.tensor([1,2,3,4,5,6])

y= x.reshape(2,3)
print(x, y, sep='\n')

z= x.reshape(1,6)
print(z)

z1= x.reshape(6,1)
print(z1)

x = torch.tensor([1, 2, 3, 4, 5, 6, 7, 8])

print(x.reshape(2, 4))
print(x.reshape(4, 2))
print(x.reshape(1, 8))
```

```python
x= torch.tensor([1,2,3,4])

y= x.view(2,2)

y[0,0]=100

print(y)
print(x)

x = torch.tensor([10, 20, 30, 40])

y = x.view(2, 2)

y[1, 0] = 99

print("y:")
print(y)

print("x:")
print(x)
```

```python
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print(x.shape)

y = x.flatten()

print(y)
print(y.shape)
```

```python
x= torch.tensor([1,2,3])
print(x.shape)
print(x)

y= x.unsqueeze(0)

print(y)
print(y.shape)

x = torch.tensor([10, 20, 30])

a = x.unsqueeze(0)
b = x.unsqueeze(1)

print(a)
print(a.shape)

print(b)
print(b.shape)
```
