```python
import pandas as pd
```

```python
data = [
    ["ayush", 22, 85, "Delhi"],
    ["rahul", 24, 91, "Mumbai"],
    ["aman", 21, 78, "Jaipur"],
    ["rohit", 23, 88, "Delhi"],
    ["neha", 25, 95, "Mumbai"]
]

df = pd.DataFrame(
    data,
    columns=["name", "age", "marks", "city"]
)
```

```python
df.sort_values("marks")
```

```python
df.sort_values("marks", ascending=False)
```

```python
df.sort_values(
    ["city", "marks"],
    ascending=[True, False]
)
```

```python
df.sort_index(ascending=False)
```

```python
df.sort_values("name", inplace= True)
print(df)
```

```python
import pandas as pd

data = [
    ["ayush", 22, 85, "Delhi"],
    ["rahul", 24, 91, "Mumbai"],
    ["aman", 21, 78, "Jaipur"],
    ["rohit", 23, 88, "Delhi"],
    ["neha", 25, 95, "Mumbai"]
]

df = pd.DataFrame(
    data,
    columns=["name", "age", "marks", "city"]
)

df.sort_values("marks", ascending=False)
```

```python
df.sort_values("marks", ascending=False).head(3)
```

```python
df.sort_index()
```

```python
df[df["marks"]>80].sort_values(
    "marks", ascending= False
)
```

```python
df[df["city"]=="Delhi"].sort_values(
    "marks", ascending= False
).head(2)
```

```python
df.sort_values("marks", ascending= False).head(2)
```

```python
df[df["age"]>22].sort_values(
    "marks", ascending= False
)
```

```python
df.sort_values(
    ["city", "age"], ascending=[True, False]
)
```
