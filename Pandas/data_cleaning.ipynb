```python
data = [
    [" Ayush ", "22", "₹40,000", "Delhi"],
    ["RAHUL", "24", "₹55,000", "delhi "],
    [" Aman ", "abc", "₹45,000", "DELHI"],
    ["Neha", "25", "₹60,000", " Mumbai"]
]

df = pd.DataFrame(
    data,
    columns=["name", "age", "salary", "city"]
)
df
```

```python
df["name"]=df["name"].str.strip()

df["city"]=(
    df["city"]
    .str.strip()
    .str.lower()
)   
df
```

```python
df["age"]=pd.to_numeric(
    df["age"],
    errors="coerce"
)
df
```

```python
df["salary"]=(
    df["salary"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
)
df["salary"]= pd.to_numeric(df["salary"])
df
```
