```python
import numpy as np
import pandas as pd
```

```python
data=[
    ["ayush", 22, 85, "Delhi"],
    ["rahul", None, 91, "mumbai"],
    ["aman", 21, np.nan, "jaipur"],
    ["rohit", 23, 88, None]
]

df= pd.DataFrame(
    data,
    columns=["name","age","marks","city"]
)

print(df)
```

```python
print(df.isna(), '\n')
print(df.isnull())
```

```python
print(df.isnull().sum(), '\n')
print(df["age"].isnull().any(), '\n')
print(df.dropna(), '\n')
print(df.dropna(subset=["marks"]), '\n')

df["age"]= df["age"].fillna(0)
print(df, '\n')
```

```python
df["age"]= df["age"].fillna(df["age"].mean())
print(df)
```

```python
df["age"]= df["age"].fillna(df["age"].median())
print(df)
```

```python
df["city"]= df["city"].fillna(df["city"].mode()[0])
print(df)
```

```python
df["city"]= df["city"].ffill()
print(df)
```

```python
df["city"]= df["city"].bfill() #uses next available value
print(df)
```
