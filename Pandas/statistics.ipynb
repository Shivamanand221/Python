```python
import pandas as pd
```

```python
data = [
    ["Ayush", 22, 85],
    ["Rahul", 24, 91],
    ["Aman", 21, 78],
    ["Rohit", 23, 88],
    ["Neha", 25, 95]
]

df = pd.DataFrame(data, columns=["name", "age", "marks"])
```

```python
print(df["marks"].mean())
print(df["marks"].sum())
print(df["marks"].min())
print(df["marks"].max())
print(df["marks"].median())
```

```python
df["marks"].describe()
```

```python
df[["age", "marks"]].mean()
```

```python
df[["age", "marks"]].describe()
```

```python
df[["name"]].describe()
```

```python
df["marks"].agg(["min","max","mean"])
```

```python
df.agg({
    "age":["min", "max", "mean"],
    "marks":["min", "max", "mean"]
})
```

```python
data = [
    ["Ayush", 22, 85000],
    ["Rahul", 24, 91000],
    ["Aman", 21, 78000],
    ["Rohit", 23, 88000],
    ["Neha", 25, 95000]
]

df = pd.DataFrame(data, columns=["name", "age", "salary"])

df
```

```python
print(df["salary"].mean())
print(df["salary"].max())
print(df["age"].min())
print(df["salary"].median())
print(df["salary"].agg(["mean", "min", "max"]))
print(df[["age", "salary"]].describe())
```
