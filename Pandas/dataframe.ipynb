```python
import pandas as pd
import numpy as np
```

```python
data2 = [
    ['ayush', 22, 'delhi'],
    ['rahul', 24, 'mumbai'],
    ['aman', 21, 'jaipur']
]

df = pd.DataFrame(data2, columns=['name', 'age', 'city'])
print(df)
```

```python
data = {
    'name': ['rahul', 'aman', 'priya'],
    'age': [22, 25, 24],
    'salary': [40000, 55000,50000]
}

df = pd.DataFrame(data)
print(
    data,
    df,
    df['name'],
    df['age'],
    df.shape,
    df.columns,
    df.index,
    df.dtypes,
    df.head(2),
    df.tail(2),
    df.info(),
    df.describe(),
    sep='\n'
)
```
