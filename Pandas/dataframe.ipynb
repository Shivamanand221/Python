```python
import pandas as pd
import numpy as np
```

```python
data2 = [
    ['ayush', 22, 'delhi'],
    ['rahul', 24, 'mumbai'],
    ['aman', 21, 'jaipur']
]

df = pd.DataFrame(data2, columns=['name', 'age', 'city'])
print(df)
```

```python
data = {
    'name': ['rahul', 'aman', 'priya'],
    'age': [22, 25, 24],
    'salary': [40000, 55000,50000]
}

df = pd.DataFrame(data)
print(
    data,
    df,
    df['name'],
    df['age'],
    df.shape,
    df.columns,
    df.index,
    df.dtypes,
    df.head(2),
    df.tail(2),
    df.info(),
    df.describe(),
    sep='\n'
)
```

```python
data = {
    'name': ['ayush','rahul','aman'],
    'age' : [22, 24, 21],
    'city': ['delhi', 'mumbai', 'jaipur']
}

df1 = pd.DataFrame(data)
print(
    df1,
    df1.shape,
    df1.columns,
    df1.dtypes,
    df.head(2),
    sep='\n'
)
```

```python
s1 = pd.Series(['ayush', 'rahul', 'aman'])
s2 = pd.Series([22, 24, 21])

df3 = pd.DataFrame({
    'names': s1,
    'ages': s2
})

print(
    s1,
    s2,
    df3,
    sep='\n'
)
```
