```python
import pandas as pd
```

```python
data = [
    ["ayush", 22, 85],
    ["rahul", 24, 91],
    ["aman", 21, 78],
    ["rohit", 23, 88]
]

df = pd.DataFrame(
    data,
    columns=["name", "age", "marks"]
)

print(
    df["name"],
    df.name,
    df[["name", "marks"]],
    df.iloc[0],
    df.iloc[0:2],
    df.loc[0],
    sep='\n'
)

df.index = ["A", "B", "C", "D"]
print(
    df,
    df.loc["B"],
    df.iloc[1],
    sep='\n'
)
```

```python
data = [
    ["ayush", 22, 85],
    ["rahul", 24, 91],
    ["aman", 21, 78],
    ["rohit", 23, 88]
]

df = pd.DataFrame(
    data,
    columns=["name", "age", "marks"]
)

df.index = ["A", "B", "C", "D"]

print(
    df.loc["B", "marks"],
    df.loc[["A","C"], ["name", "marks"]],
    df.iloc[1,2],
    df.iloc[0:3, 0:2], #3 and 2 excluded
    df.loc["A":"C", "name":"age"], #C included
    df[df["marks"]>85],
    df.at["B", "marks"],
    df.iat[1,2],
    sep='\n'
)
```

```python
import pandas as pd

data = [
    ["ayush", 22, 85, "Delhi"],
    ["rahul", 24, 91, "Mumbai"],
    ["aman", 21, 78, "Jaipur"],
    ["rohit", 23, 88, "Delhi"],
    ["neha", 25, 95, "Mumbai"]
]

df = pd.DataFrame(
    data,
    columns=["name", "age", "marks", "city"]
)

print(
    df[df["marks"] > 85],
    df[df["age"] <= 23],
    df[df["age"] != 23],
    df[df["city"] == "Delhi"],
    sep='\n'
)
```

```python
print(
    df[(df["marks"]>85) & (df["age"]>22)],
    df[~(df["city"] == "Delhi")], #~ represents NOT
    df[(df["city"] == "Delhi") | (df["city"] == "Mumbai")],
    df[df["city"].isin(["Delhi", "Mumbai"])],
    df[df["marks"].between(80, 90)],
    sep='\n'
)
```

```python
print(
    df[
        (df["marks"]>80)&
        (df["age"].between(22,24))&
        (df["city"].isin(["Delhi", "Mumbai"]))
    ]
)

print(
    df.loc[
        df["marks"]>85,
        ["name", "marks"]
    ]
)

print(df.query("marks>85"))
print(df.query("marks>85 and age>22"))
print(df[(df["marks"] > 85) & (df["age"] > 22)]==df.query("marks > 85 and age > 22"))
```
