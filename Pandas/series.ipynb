```python
import pandas as pd
import numpy as np
```

```python
s = pd.Series([10, 20, 30, 40])
print(s)
```

```python
arr = np.array([10, 20, 30, 40])
s1 = pd.Series(arr)
print(arr, s1, sep='\n')
```

```python
data = {
    "math": 90,
    "physics": 85,
    "chemistry": 92
}

s2 = pd.Series(data)
print(data, s2, sep='\n')
```

```python
s3= pd.Series(
    [90, 80, 70],
    index= ['math','physics','chemistry']
)
print(s3)
print(s3['math'], s3['physics']) #access by label
print(s3.iloc[1]) #access by position
print(s3.values)
print(s3.index)
print(s3.dtype)
print(s3.shape)
```

```python
s4 = pd.Series([10, 20, 30, 40])
print(s4)
print(s4+10)
print(s4*2)
print(s4/10)
print(s4**2)
```

```python
s5 = pd.Series([1, 2, 3, 4])
print(s4+s5)
```

```python
s6 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
s7 = pd.Series([1, 2, 3], index=['b', 'c', 'd'])
print(s6+s7) # NaN for non-matching indices
```

```python
s8= pd.Series([10, 20, 30, 40])

print(
    s8.sum(),
    s8.count(),
    s8.max(),
    s8.min(),
    s8.mean(),
    s8.median(),
    s8.std(),
    sep='\n'
)
```
