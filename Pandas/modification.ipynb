```python
data= [
    ["ayush", 22, 85, "delhi"],
    ["rahul", 24, 91, "Mumbai"],
    ["aman", 21, 78, "Jaipur"],
    ["rohit", 23, 88, "Delhi"],
    ["neha", 25, 95, "Mumbai"]
]

df= pd.DataFrame(
    data,
    columns=["name", "age", "marks", "city"]
)

new_rows= pd.DataFrame([
    ["karan", 26, 89, "Pune"],
    ["riya", 22, 93, "Delhi"]
], columns=["name", "age", "marks", "city"])

df = pd.concat([df, new_rows], ignore_index=True)

print(df)
```

```python
df=df[df["marks"]>=80]
print(df)
```

```python
data = [
    ["ayush", 22, 85, "Delhi"],
    ["rahul", 24, 91, "Mumbai"],
    ["aman", 21, 78, "Jaipur"],
    ["rohit", 23, 88, "Delhi"]
]

df = pd.DataFrame(
    data,
    columns=["name", "age", "marks", "city"]
)

df["passed"] = df["marks"] >= 80

df["age_after_10"] = df["age"] + 10

df.drop("city", axis=1, inplace=True)

df.drop(2, inplace=True)

df.loc[4] = ["neha", 25, 95, True, 35]

print(df)
```
