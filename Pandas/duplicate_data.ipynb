```python
import pandas as pd

data = [
    ["ayush", 22, 85, "Delhi"],
    ["rahul", 24, 91, "Mumbai"],
    ["aman", 21, 78, "Jaipur"],
    ["rahul", 24, 91, "Mumbai"],
    ["rohit", 23, 88, "Delhi"],
    ["aman", 21, 78, "Jaipur"]
]

df = pd.DataFrame(
    data,
    columns=["name", "age", "marks", "city"]
)

print(df)
```

```python
print(df.duplicated(), '\n')

print(df.duplicated().sum(), '\n')

print(df[df.duplicated()])
```

```python
print(df.drop_duplicates(), '\n')
print(df, '\n')

df.drop_duplicates(inplace=True)
print(df, '\n')
```
