```python
import pandas as pd
```

```python
data = [
    ["Ayush", "IT", 50000],
    ["Rahul", "HR", 45000],
    ["Aman", "IT", 60000],
    ["Rohit", "Sales", 40000],
    ["Neha", "HR", 55000],
    ["Priya", "Sales", 48000]
]

df = pd.DataFrame(data, columns=["name", "department", "salary"])
df
```

```python
print(df.groupby("department")["salary"].mean())
print(df.groupby("department")["salary"].sum())
print(df.groupby("department")["salary"].max())
```
